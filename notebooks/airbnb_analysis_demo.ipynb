## 4. Advanced Analysis and Insights

### 4.1 Price Elasticity by Season

```python
# Analyze price elasticity across seasons
seasonal_data_copy = seasonal_data.copy()
seasonal_data_copy['month'] = pd.to_datetime(seasonal_data_copy['date']).dt.month

# Define seasons
def get_season(month):
    if month in [12, 1, 2]:
        return 'Summer'
    elif month in [3, 4, 5]:
        return 'Autumn'
    elif month in [6, 7, 8]:
        return 'Winter'
    else:
        return 'Spring'

seasonal_data_copy['season'] = seasonal_data_copy['month'].apply(get_season)

# Price sensitivity analysis
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
seasons = ['Summer', 'Autumn', 'Winter', 'Spring']

for i, season in enumerate(seasons):
    row, col = i // 2, i % 2
    season_data = seasonal_data_copy[seasonal_data_copy['season'] == season]
    
    # Price vs accommodates for each season
    sns.scatterplot(data=season_data, x='accommodates', y='price', 
                   hue='room_type', alpha=0.6, ax=axes[row, col])
    axes[row, col].set_title(f'{season} - Price vs Accommodates')
    
    # Add trend line
    z = np.polyfit(season_data['accommodates'], season_data['price'], 1)
    p = np.poly1d(z)
    axes[row, col].plot(season_data['accommodates'], p(season_data['accommodates']), 
                       "r--", alpha=0.8, linewidth=2)

plt.tight_layout()
plt.show()

# Save the visualization
fig.savefig('../results/visualizations/seasonal_price_elasticity.png', 
           dpi=300, bbox_inches='tight')
print("💾 Seasonal price elasticity analysis saved to results/visualizations/seasonal_price_elasticity.png")
```

### 4.2 Market Segmentation Analysis

```python
# Market segmentation by price ranges
seasonal_data_copy['price_segment'] = pd.cut(seasonal_data_copy['price'], 
                                           bins=[0, 75, 125, 200, float('inf')],
                                           labels=['Budget', 'Mid-range', 'Premium', 'Luxury'])

# Segment analysis by season
segment_analysis = seasonal_data_copy.groupby(['season', 'price_segment']).agg({
    'price': ['count', 'mean'],
    'accommodates': 'mean',
    'availability_365': 'mean'
}).round(2)

print("🎯 Market Segmentation by Season:")
print(segment_analysis)

# Visualize market segments
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Segment distribution by season
segment_counts = seasonal_data_copy.groupby(['season', 'price_segment']).size().unstack()
segment_counts.plot(kind='bar', stacked=True, ax=ax1)
ax1.set_title('Market Segment Distribution by Season')
ax1.set_xlabel('Season')
ax1.set_ylabel('Number of Listings')
ax1.legend(title='Price Segment')
ax1.tick_params(axis='x', rotation=45)

# Average price by segment and season
avg_prices = seasonal_data_copy.groupby(['season', 'price_segment'])['price'].mean().unstack()
avg_prices.plot(kind='bar', ax=ax2)
ax2.set_title('Average Price by Segment and Season')
ax2.set_xlabel('Season')
ax2.set_ylabel('Average Price ($)')
ax2.legend(title='Price Segment')
ax2.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# Save the visualization
fig.savefig('../results/visualizations/market_segmentation.png', 
           dpi=300, bbox_inches='tight')
print("💾 Market segmentation analysis saved to results/visualizations/market_segmentation.png")
```
